```python
import pandas as pd

# Crear un DataFrame de ejemplo
data = {
    'Departamento': ['Ventas', 'Ventas', 'IT', 'IT', 'RRHH', 'RRHH', 'Ventas'],
    'Empleado': ['Ana', 'Luis', 'Pedroy ', 'Marta', 'Sofía', 'Carlos', 'David'],
    'Salario': [30000, 50000, 60000, 55000, 40000, 42000, 48000],
    'Edad': [25, 30, 40, 35, 28, 45, 32]
}
df = pd.DataFrame(data)

# Lista de nombres femeninos (puede mejorarse con un análisis de datos)
nombres_femeninos = ['Ana', 'Marta', 'Sofía']

# Asignar el valor de sexo usando groupby y apply
df['Sexo'] = df.groupby('Departamento')['Empleado'].transform(lambda x: x.apply(lambda nombre: 1 if nombre in nombres_femeninos else 0))

#print(df)

df
```
