```python
data1 = pd.DataFrame({
    'ID': [1,2,3,4],
    'Nombre': ['Ana', 'Luis', 'Juan', 'Alberto']
})

data2 = pd.DataFrame({
    'ID': [3,4,5],
    'Salario': ['20000', '25000', '30000']
})

# Fusión de ambos DataFrames por la columna 'ID' (común)
data_merged_inner = pd.merge(data1, data2, how='inner', on='ID')

print(data_merged_inner, "\n")

data_merged_outer = pd.merge(data1, data2, how='outer', on='ID')

print(data_merged_outer)
#groupById = data_merged.groupby('ID')

#print(groupById)
```

```python
df1 = pd.DataFrame({
    'ID_cliente': [1,2,3,4,5]
})

df2 = pd.DataFrame({
    'ID_usuario': [1,2,3,4,5]
})

df_merged = pd.merge(df1, df2, how='inner', left_on='ID_cliente', right_on='ID_usuario')
```

```python
from tabulate import tabulate

print(tabulate(df_merged, headers=["ID_cliente", "ID_usuario"], tablefmt='orgtbl'))
```

```python
from sklearn.datasets import load_breast_cancer

# Cargar el dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)

# Añadir la columna 'target' (maligno o benigno)
df['target'] = data.target

# Mostrar las primeras filas para ver cómo está estructurado el DataFrame

df_modified = df['mean radius'] + df['mean texture']

df_final = pd.DataFrame(df_modified)

print(tabulate(df_final, headers=["mean radius", "mean tecture"], tablefmt='orgtbl'))
```
